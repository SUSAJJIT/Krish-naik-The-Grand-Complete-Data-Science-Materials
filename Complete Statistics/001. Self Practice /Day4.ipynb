```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
##define out dataset
dataset= [11,10, 12, 14, 12, 15, 14,13,15,102,12,14,17,19,107, 10,13,12,14,12,108,12,11,14,13,15,10,15,12,10,14,13,15,10]
```

```python
## Z score and outlier

outliers =[]


def detect_outlier(dataset):
  threshold=3
  mean = np.mean(dataset)
  sd =np.std(dataset) # sd = standard deviation

  for i in dataset:
    z_score = (i-mean)/sd

    if np.abs(z_score)>3:
      outliers.append(i)

  return outliers
```

```python
detect_outlier(dataset)
```

## IQR --inter quartile range

steps to calculate IQR :



1.   sort the data
2.   calculate q1 and q3
1.   iqr = q3-q1
2.   find the lower fence (q1-1.5*iqr)
1.   find the upper fence (q3 + 1.5*iqr)



```python
dataset = sorted(dataset)
```

```python
# finding q1 and q3
q1,q3 = np.percentile(dataset,[25,75])
```

```python
print(q1,q3)
```

```python
## finding IQR
iqr = q3-q1
print(iqr)
```

```python
#finding lower and higher fence

lower_fence = q1 - (1.5*iqr)
upper_fence = q3 + (1.5*iqr)

# most of the elements will be bw these 2 values
print(lower_fence,upper_fence)
```
